## Find index of max / min value of a column

- `.idxmin`: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.idxmin.html

- `.idxmax`: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.idxmax.html

```python
%config InteractiveShell.ast_node_interactivity="all"

from sklearn.datasets import load_diabetes

data = load_diabetes(as_frame=True)
df = data.frame

max_age = df['age'].max()
max_age_index_1 = df[df['age'] == max_age].index
max_age_index_1

max_age_index_2 = df['age'].idxmax()
max_age_index_2

min_age_index = df['age'].idxmin()
min_age_index
```

## Find n smallest / largest values of a column

- `.nsmallest`: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.nsmallest.html

- `.nlargest`: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.nlargest.html

```python
%config InteractiveShell.ast_node_interactivity="all"

from sklearn.datasets import load_breast_cancer

data = load_breast_cancer(as_frame=True)
df = data.frame

df.nsmallest(3, 'mean radius')
df.nlargest(2, 'mean radius')
```

## Find not equal

- `.ne`: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.ne.html

```python
import pandas as pd

df = pd.DataFrame({'X': [0,0,0,0,0,0,0,1,2,3,4,5]})
mask = df['X'].ne(0)
df[mask]
```

## Convert DataFrame to Markdown

- `.to_markdown`: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_markdown.html

```python
from sklearn.datasets import load_iris

data = load_iris(as_frame=True)
df = data.frame

print(df.to_markdown())
```
