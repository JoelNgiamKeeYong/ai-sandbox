# **🧪 Python Libraries**

## **└─ Scikit-learn**

- Datasets API: https://scikit-learn.org/stable/api/sklearn.datasets.html

- Provides toy datasets (e.g., Iris, Wine, Digits) for quick experimentation and demos.

- Includes dataset fetchers (e.g., fetch_openml, fetch_california_housing) to load larger, real-world datasets.

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing(as_frame=True)
X = data.data
y = data.target

df = pd.concat([X, y.rename('target')], axis=1)
df.head()
```

## **└─ TensorFlow**

- Datasets API: https://www.tensorflow.org/datasets

- Offers a wide collection of ready-to-use datasets for machine learning.

- Covers images, text, audio, and structured data, preprocessed and split into train/test sets for direct use with TensorFlow or other ML frameworks.

---
---
## **└─ PyTorch**

- `TorchVision`, `TorchText`, and `TorchAudio` provide domain-specific datasets (e.g., CIFAR-10, COCO for vision; IMDB for text; LibriSpeech for audio).

- `TorchVision` Datasets API: https://docs.pytorch.org/vision/stable/datasets.html

- `TorchText` Datasets API: https://docs.pytorch.org/text/stable/datasets.html

- `TorchAudio` Datasets API: https://docs.pytorch.org/audio/stable/datasets.html

- They integrate seamlessly with PyTorch’s DataLoader, making it easy to prototype deep learning models on benchmark datasets.

---
---
## **└─ Seaborn**

- Datasets API: https://seaborn.pydata.org/generated/seaborn.load_dataset.html

- Comes with several small, clean datasets (e.g., Titanic, Tips, Penguins) accessible via `seaborn.load_dataset()`.

- Designed mainly for data visualization demos, but handy for practice in exploratory data analysis (EDA).

---
---
## **└─ Statsmodels**

- Datasets API: https://www.statsmodels.org/stable/datasets/index.html

- Includes a set of classic statistical datasets (e.g., Longley, Star98, Grunfeld) for regression, time series, and econometrics.

- Useful for teaching, reproducing textbook examples, and testing statistical methods.

- Datasets API: https://www.openml.org/
