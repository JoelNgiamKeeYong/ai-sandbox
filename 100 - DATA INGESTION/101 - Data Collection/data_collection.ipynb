# **📁 Data Storage File Types**

- Used for storing datasets that are loaded into pandas, NumPy, Spark, etc.

| File Type              | Description                                 | When to Use                                  |
| ---------------------- | ------------------------------------------- | -------------------------------------------- |
| **`.csv`**             | Plain text, comma-separated values          | ✅ Small to medium datasets<br>✅ Easy to edit |
| **`.json`**            | Structured, hierarchical (dict-like) data   | ✅ Nested or semi-structured data             |
| **`.xlsx`**            | Excel spreadsheet                           | ✅ Business data or user-supplied files       |
| **`.parquet`**         | Columnar, compressed binary format          | ✅ Big data<br>✅ Fast I/O<br>✅ Cloud systems  |
| **`.feather`**         | Lightweight columnar format (like Parquet)  | ✅ Extremely fast in-memory read/write        |
| **`.hdf5`**            | Hierarchical binary data                    | ✅ Scientific computing, time series          |
| **`.npy` / `.npz`**    | NumPy binary formats                        | ✅ Fast model input/output                    |
| **`.pkl` / `.joblib`** | Serialized Python objects (incl. ML models) | ✅ Storing models or preprocessed data        |

## **└── Parquet from local file**

- A Parquet file is an open-source, columnar data file format designed for efficient data storage and retrieval, particularly in big data environments.
- Unlike traditional row-based formats like CSV or JSON, Parquet organizes data by columns, offering significant advantages for analytical workloads.

```python
import pandas as pd 

df = pd.read_parquet("./data/titanic.parquet")
df.head()
```

# **🛢️ Database File Types**

- Actual databases or database dumps used before or during model training.

| File Type                  | Description                           | When to Use                        |
| -------------------------- | ------------------------------------- | ---------------------------------- |
| **`.db` / `.sqlite`**      | Local SQLite database                 | ✅ Lightweight DB for ML demos      |
| **`.sql`**                 | SQL dump (PostgreSQL, MySQL, etc.)    | ✅ Reconstruct database from backup |
| **No file (connected DB)** | Cloud DBs (BigQuery, Snowflake, etc.) | ✅ Real-time or big data querying   |

## **└── .db file from local file**

- `sqlite3` comes built-in with Python.
- Limited to SQLite databases only.
- Provides a low-level, direct interface to SQLite databases.
- You write raw SQL queries and manage connections/cursors yourself.
- Good for simple scripts or quick prototyping.

```python
import sqlite3
import pandas as pd

db_path = "./data/bmarket.db"

with sqlite3.connect(db_path) as conn:
    query = "SELECT name FROM sqlite_master WHERE type='table';"
    tables_df = pd.read_sql(query, conn)
    
tables_df
```

```python
import sqlite3
import pandas as pd

db_path = "./data/bmarket.db"

with sqlite3.connect(db_path) as conn:
    query = "SELECT * from bank_marketing"
    df = pd.read_sql(query, conn)

df.head()    
```
