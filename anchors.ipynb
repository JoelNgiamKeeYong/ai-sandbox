**🧭 1. Problem Framing & Goal Alignment**

_Understand the context and what you’re solving._

| Prompt          | What It Triggers You to Ask                           |
| --------------- | ----------------------------------------------------- |
| **Goal**        | What’s the objective? What does success look like?    |
| **Problem**     | What’s the pain point or business need?               |
| **Stakeholder** | Who are we solving this for? What do they care about? |
| **Constraint**  | What are the limits (time, data, budget, compliance)? |


**📊 2. Data Understanding & Feasibility**

_Evaluate the inputs, risks, and limitations._

| Prompt       | What It Triggers You to Ask                                    |
| ------------ | -------------------------------------------------------------- |
| **Data**     | What kind of data do we have? Is it labeled, complete, usable? |
| **Bias**     | Are there risks of unfairness or skewed outcomes?              |
| **Baseline** | Can we solve this with a simple non-AI or rule-based method?   |

**🧠 3. Solution Design & Modeling**

_Brainstorm and refine possible technical approaches._

| Prompt        | What It Triggers You to Ask                                          |
| ------------- | -------------------------------------------------------------------- |
| **Model**     | What ML approach fits best (classification, regression, clustering)? |
| **Metrics**   | How do we measure success (F1, AUC, ROI)? Do we monitor drift?       |
| **Trade-off** | Are we optimizing for accuracy, speed, cost, interpretability?       |
| **Risk**      | What are the costs of being wrong (false positives/negatives)?       |

**🚀 4. Deployment & Scalability**

_Think long term — how the solution runs and grows._

| Prompt      | What It Triggers You to Ask                                    |
| ----------- | -------------------------------------------------------------- |
| **Scale**   | Can this work as data or users increase? Is it robust?         |
| **Deploy**  | How do we roll this out? What does production look like?       |
| **Performance** | Do we monitor drift? |


**🤝 5. Teamwork & Decision Making**

_Show collaboration, structure, and leadership._

| Prompt      | What It Triggers You to Ask                                   |
| ----------- | ------------------------------------------------------------- |
| **Team**    | Are we including everyone? Who can summarize where we are?    |
| **Timebox** | Are we managing time effectively? Do we need to move forward? |
| **Next**    | What’s our chosen solution? What would we do if given a week? |
