| Encoding Type | Use When                      | Pros                       | Cons                         |
| ------------- | ----------------------------- | -------------------------- | ---------------------------- |
| One-Hot       | Low cardinality, no order     | Simple, model-friendly     | Feature explosion            |
| Ordinal       | Natural order in categories   | Keeps order                | Misleading if order is wrong |
| Target        | Tree models, high cardinality | Encodes signal from target | Risk of data leakage         |
| Frequency     | Any model, low risk           | Simple, robust             | No target signal             |
| Binary        | Medium to high cardinality    | Compact                    | Less interpretable           |
| Hashing       | Very high cardinality         | Fixed-dimension, fast      | No interpretability          |
