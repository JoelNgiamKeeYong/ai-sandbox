# **✂️ Data Splitting**

- Data splitting is a fundamental step in preparing your dataset before training machine learning models.

- This process involves dividing your data into separate sets, typically known as the training set, validation set, and testing set.

- This is also known as train-validation-test data splitting.

- Each set serves a unique purpose in the machine learning pipeline, ensuring that your model generalises well to unseen data.
