# Regression Metrics

---
## Metric Behavior Cheatsheet

| **Metric**        | **Penalizes Outliers** | **Interpretability**            | **Scale-Sensitive** | **Handles % Error** | **Works with Skewed Data** | **Sensitive to Small Errors** | **Unit of Measure**       | **Robust to Outliers** | **Notes**                                  |
| ----------------- | ---------------------- | ------------------------------- | ------------------- | ------------------- | -------------------------- | ----------------------------- | ------------------------- | ---------------------- | ------------------------------------------ |
| **MAE**           | ❌ No                   | ✅ High (same units as target)   | ❌ No                | ❌ No                | ❌ No                       | ✅ Yes                         | Same as target            | ✅ Yes                  | Intuitive average error                    |
| **MSE**           | ✅ Yes (strongly)       | ❌ Low (squared units)           | ✅ Yes               | ❌ No                | ❌ No                       | ❌ No                          | Squared target units      | ❌ No                   | Overemphasizes large errors                |
| **RMSE**          | ✅ Yes (strongly)       | ✅ Moderate                      | ✅ Yes               | ❌ No                | ❌ No                       | ❌ No                          | Same as target            | ❌ No                   | Like MAE but more punishing                |
| **MedAE**         | ❌ No                   | ✅ Very High (median error)      | ❌ No                | ❌ No                | ✅ Yes                      | ✅ Yes                         | Same as target            | ✅ Yes                  | Robust version of MAE                      |
| **R² Score**      | ❌ No                   | ✅ Explains variance (%)         | ❌ No                | ❌ No                | ❌ No                       | ❌ No                          | Unitless (0–1 or -∞ to 1) | ❌ No                   | Not valid for some non-linear or poor fits |
| **Adjusted R²**   | ❌ No                   | ✅ (adjusts for # features)      | ❌ No                | ❌ No                | ❌ No                       | ❌ No                          | Unitless                  | ❌ No                   | Better for comparing models                |
| **MAPE**          | ❌ No                   | ✅ Extremely High (% error)      | ✅ Yes               | ✅ Yes               | ❌ No (breaks on 0)         | ✅ Yes                         | Percentage (%)            | ❌ No                   | Easy for stakeholders; avoid 0 targets     |
| **MSLE**          | ✅ (mildly)             | ✅ Good for growth trends        | ✅ Yes (log scale)   | ❌ No                | ✅ Yes (log handles skew)   | ❌ No                          | Log units                 | ✅ Yes (somewhat)       | Useful for exponential targets             |
| **SMAPE**         | ❌ (mildly)             | ✅ Better than MAPE for 0s       | ✅ Yes               | ✅ Yes               | ✅ Partial                  | ✅ Yes                         | Percentage (%)            | ✅ Yes (somewhat)       | Symmetric version of MAPE                  |
| **Explained Var** | ❌ No                   | ✅ How much variance is captured | ❌ No                | ❌ No                | ❌ No                       | ❌ No                          | Unitless                  | ❌ No                   | Similar to R², less commonly used          |

- MAE = Simple, interpretable, robust.
- RMSE = Use when large errors are costly.
- MAPE/SMAPE = Great for % understanding — avoid MAPE if values can be 0.
- MSLE = Excellent for log-normal targets, e.g., income or growth.
- R² / Adjusted R² = Good for measuring model fit, especially for feature comparisons.

## Examples for Bank Problems

| Use Case                                               | Target Variable Characteristics                 | Recommended Metrics                   | Why                                                                                                                  |
| ------------------------------------------------------ | ----------------------------------------------- | ------------------------------------- | -------------------------------------------------------------------------------------------------------------------- |
| **🏦 Predicting loan amount / credit limit**           | Monetary, continuous, can vary widely           | **MAE**, **RMSE**, **R²**             | - RMSE penalizes large mistakes (high-stake)<br>- MAE is interpretable (\$-based)<br>- R² for explaining performance |
| **📊 Predicting number of transactions / purchases**   | Count data, non-negative integers, right-skewed | **MAE**, **RMSE**, **MSLE**           | - MSLE handles skew + avoids punishing small errors<br>- RMSE for large mistake sensitivity                          |
| **📈 Predicting customer lifetime value (CLV)**        | Highly skewed monetary value                    | **MSLE**, **MAE**, **R²**             | - MSLE better for log-normal targets<br>- MAE to show expected dollar error                                          |
| **🕒 Time-based prediction (e.g. spending over time)** | Potentially seasonally or log-growing           | **RMSE**, **MAPE**, **R²**            | - MAPE helps compare across time windows<br>- RMSE for bigger mistakes                                               |
| **🧾 Predicting account balance**                      | Continuous, strictly positive                   | **MAE**, **RMSE**, **MSLE**           | - MSLE reduces over-penalizing outliers<br>- MAE for simplicity                                                      |
| **🛒 Predicting price or demand**                      | Can have small values, sometimes 0              | **MAE**, **RMSE** *(avoid MAPE/MSLE)* | - MAPE breaks on 0<br>- Stick with MAE/RMSE                                                                          |
| **🏘 Predicting house prices / asset value**           | Wide monetary range, often right-skewed         | **RMSE**, **MAE**, **R²**             | - RMSE emphasizes high-error values<br>- MAE keeps interpretation                                                    |
| **⚠ Predicting risk score or probability**             | Bounded (e.g. 0–1 or 0–100), not monetary       | **MAE**, **MSE**, **R²**              | - MAE/MSE for general accuracy<br>- No need for MSLE/MAPE here                                                       |
