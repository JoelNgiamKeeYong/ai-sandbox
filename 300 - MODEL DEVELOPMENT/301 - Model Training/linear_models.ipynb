# 🤖🧠🛠️ **Linear Models**

```python
from sklearn.datasets import load_diabetes, load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
# For regression
X_reg, y_reg = load_diabetes(return_X_y=True)
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

# Standardize features for regression
scaler_reg = StandardScaler().fit(X_reg_train)
X_reg_train_scaled = scaler_reg.transform(X_reg_train)
X_reg_test_scaled = scaler_reg.transform(X_reg_test)
```

```python
# For classification
X_clf, y_clf = load_iris(return_X_y=True)
X_clf_train, X_clf_test, y_clf_train, y_clf_test = train_test_split(X_clf, y_clf, test_size=0.2, random_state=42)

# Standardize features for classification
scaler_clf = StandardScaler().fit(X_clf_train)
X_clf_train_scaled = scaler_clf.transform(X_clf_train)
X_clf_test_scaled = scaler_clf.transform(X_clf_test)
```

---
---

# **🔖 Regression Models**

---
---
## 🔖 **Linear Regression**

```python
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()
lr_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = lr_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ Linear Regression - MSE: {mse:.4f}")
```

---
---
## 🔖 **Ridge Regression**

```python
from sklearn.linear_model import Ridge
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = ridge_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ Ridge Regression - MSE: {mse:.4f}")
```

---

```python
from sklearn.linear_model import Lasso
lasso_model = Lasso(alpha=0.1)
lasso_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = lasso_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ Lasso Regression - MSE: {mse:.4f}")
```

## 🏷️ **Logistic Regression (lbfgs)**

- Logistic Regression hyperparameters mainly focus on regularization type and strength to control overfitting, solver choice to efficiently optimize, and handling class imbalance.

- Tuning penalty and C are often the most impactful.

**🔧 Regularization & Penalty**

| Hyperparameter | Default | What It Does                                                        | Effect on Model                                                                                    |
| -------------- | ------- | ------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------- |
| `penalty`      | 'l2'    | Type of regularization: `'l1'`, `'l2'`, `'elasticnet'`, or `'none'` | Controls model complexity and feature shrinkage to avoid overfitting                               |
| `C`            | 1.0     | Inverse of regularization strength (`C = 1/λ`)                      | Smaller `C` means stronger regularization (more shrinkage), larger `C` means weaker regularization |

**⚙️ Solver & Optimization**

| Hyperparameter | Default | What It Does                                                                         | Effect on Model                                                              |
| -------------- | ------- | ------------------------------------------------------------------------------------ | ---------------------------------------------------------------------------- |
| `solver`       | 'lbfgs' | Algorithm to optimize logistic regression (`'liblinear'`, `'lbfgs'`, `'saga'`, etc.) | Different solvers support different penalties and scale better on large data |
| `max_iter`     | 100     | Maximum number of iterations for solver                                              | Increase if solver doesn’t converge                                          |

**🧮 Additional Controls**
| Hyperparameter  | Default | What It Does                                            | Effect on Model                                                             |
| --------------- | ------- | ------------------------------------------------------- | --------------------------------------------------------------------------- |
| `class_weight`  | None    | Weights associated with classes (`'balanced'` or dict)  | Helps handle imbalanced classes by giving more importance to minority class |
| `fit_intercept` | True    | Whether to calculate the intercept term                 | Usually True; can be False if data already centered                         |
| `multi_class`   | 'auto'  | Type of multi-class strategy (`'ovr'`, `'multinomial'`) | Controls how multi-class problems are handled                               |


```python
from sklearn.linear_model import LogisticRegression  
lr_model = LogisticRegression(max_iter=1000, solver='lbfgs')
lr_model.fit(X_clf_train_scaled, y_clf_train)
y_pred = lr_model.predict(X_clf_test_scaled)
acc = accuracy_score(y_clf_test, y_pred)
print(f"Logistic Regression - Accuracy: {acc:.4f}")
```
