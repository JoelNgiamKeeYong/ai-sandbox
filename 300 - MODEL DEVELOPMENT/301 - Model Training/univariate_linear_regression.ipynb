
## Overview
In this interactive Python notebook, we delve into the fundamental concepts of Univariate Linear Regression. This activity is designed to provide hands-on experience with linear regression, allowing for an intuitive understanding of how changes in model parameters affect the regression line and the overall prediction accuracy.

Linear regression is a linear approach to modeling the relationship between a dependent variable and one or more independent variables.

In Univariate Linear Regression (i.e. a single variable $x$), our model predicts the dependent variable, $y$, as a function of a single independent variable, $x$. The relationship is modeled through a linear function:

$$
y = mx + c
$$

where:
- $y$ is the dependent variable.
- $x$ is the independent variable.
- $m$ is the slope of the line (also known as weight or coefficient).
- $c$ is the y-intercept.

## Objective
The primary objective of this activity is to understand how the slope (gradient) and y-intercept of the regression line influence the prediction and how the model fits the data. We aim to minimize the sum of squared differences between the observed values and the values predicted by the model, known as the sum of squares error.

## Interactive Visualization
The notebook includes interactive sliders for dynamically altering the slope (gradient) and y-intercept of the regression line. This interactivity allows you to observe in real-time how these changes influence the model's fit to the data.



You can also use the [scikit-learn library in Python](https://scikit-learn.org/stable/) to perform [linear regression](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html) very efficiently.

Below is the equivalent code using scikit-learn to perform the same univariate linear regression as before.


This code also includes the plotting of the regression line and the interactive sliders for gradient and intercept, although please note that scikit-learn will compute its own optimal values for these parameters.

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from ipywidgets import interact

# Generate some random data points
np.random.seed(0)
num_pts = 5
x = 2 * np.random.rand(num_pts, 1)
y = 4 + 3 * x + np.random.randn(num_pts, 1)

# Fit the model using Scikit-Learn
lin_reg = LinearRegression()
lin_reg.fit(x, y)
intercept, slope = lin_reg.intercept_[0], lin_reg.coef_[0][0]

# Function to plot the data, regression lines, and calculate sum of squares error
def plot_regression_line(m, c):
    plt.scatter(x, y)

    # Plotting the line from interactive parameters
    y_pred = m * x + c
    plt.plot(x, y_pred, color='red', label='Interactive Line')

    # Plotting the Scikit-Learn regression line
    y_pred_sklearn = slope * x + intercept
    plt.plot(x, y_pred_sklearn, color='green', label='Least square regression Line')

    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Univariate Linear Regression')
    plt.legend()
    plt.show()

    # Calculate and print sum of squares error for the interactive line
    ss_error_interactive = np.sum((y - y_pred) ** 2)
    print(f"Sum of squares error (Interactive Line): {ss_error_interactive}")

    # Calculate and print sum of squares error for the Scikit-Learn line
    ss_error_sklearn = np.sum((y - y_pred_sklearn) ** 2)
    print(f"Least Sum of squares error (Scikit-Learn Line): {ss_error_sklearn}")

    # Display the optimal gradient and intercept for Scikit-Learn model
    print(f"Scikit-Learn Line Gradient (Slope) m*: {slope}")
    print(f"Scikit-Learn Line Intercept c*: {intercept}")

    print(f"Therefore, optimal linear regression with sum of least squares error is : y={slope}x+{intercept}")

# Create interactive sliders for slope (m) and intercept (c)
interact(plot_regression_line, m=(-10, 10, 0.1), c=(-10, 10, 0.1));
```

## (Optional) Mathematical Explanation behind Linear Regression
The optimal parameters for our linear regression model are computed using the least squares method, minimizing the sum of squared residuals. The parameters are determined by the following equations:

1. For the least squares regression line parameters (slope $m$ and intercept $c$):
$$
\theta = (X^TX)^{-1}X^Ty
$$
where $\theta$ represents the parameter vector containing the intercept and slope, $X$ is the matrix of input features, and $y$ is the vector of observed values, provided $X$ has linearly independent columns.

Let us express the matrices $ X $  and $ y$ in more detail in the context of univariate linear regression.

1. Matrix $ X$  (also known as the feature matrix or design matrix):
In univariate linear regression, we usually add a column of ones to the matrix $X$ to account for the intercept term. If you have $n$ data points, $X$ will be an $n \times 2$ matrix where the first column is all ones (representing the intercept), and the second column contains all the $x_i$ values (the independent variable).

$$
X = \begin{pmatrix}
1 & x_{1} \\
1 & x_{2} \\
\vdots & \vdots \\
1 & x_{n}
\end{pmatrix}
$$

2. Vector $ y $ (also known as the target vector or response vector):
This is a column vector of the dependent variable values. If you have $ n $ data points, $ y $ will be an $ n \times  1 $ matrix (or an $ n$-dimensional vector).

$$
y = \begin{pmatrix}
y_{1} \\
y_{2} \\
\vdots \\
y_{n}
\end{pmatrix}
$$

Now, the equation for the least squares regression line parameters can be more contextually understood:

$$
\theta = (X^TX)^{-1}X^Ty
$$

Here, $\theta$ is a $ 2 \times 1 $ vector containing the intercept $ \theta_0$ and slope $ \theta_1 $:

$$
\theta = \begin{pmatrix}
\theta_{0} \\
\theta_{1}
\end{pmatrix}
$$

And the sum of square errors is minimized through:

$$
\text{min}_{\theta} \| X\theta - y\|_{2}^{2} = \sum\limits_{i=1}^{n} \left(\theta_{0} + \theta_{1}x_{i} - y_{i} \right)^{2}
$$

This establishes the connection between the data points, the design matrix \(X\), the parameter vector \(\theta\), and the result vector \(y\), providing a complete picture of how the least squares regression line parameters are calculated in univariate linear regression.


2. The sum of squares error is computed as:
$$
SSE = \sum_{i=1}^{n} (y_i - \hat{y_i})^2
$$
where $y_i$ is the actual value, $\hat{y_i}$ is the predicted value, and $n$ is the number of observations.


```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

# Generate some random data points
np.random.seed(0)
num_pts = 5
x = 2 * np.random.rand(num_pts, 1)
y = 4 + 3 * x + np.random.randn(num_pts, 1)

# Compute the parameters of the least squares line
x_b = np.c_[np.ones((num_pts, 1)), x]  # add x0 = 1 to each instance
theta_best = np.linalg.inv(x_b.T.dot(x_b)).dot(x_b.T).dot(y)

# Function to plot the data, regression lines, and calculate sum of squares error
def plot_regression_line(m, c):
    plt.scatter(x, y)

    # Plotting the line from interactive parameters
    y_pred = m * x + c
    plt.plot(x, y_pred, color='red', label='Interactive Line')

    # Plotting the least squares regression line
    y_pred_least_squares = x_b.dot(theta_best)
    plt.plot(x, y_pred_least_squares, color='blue', label='Least Squares Line')

    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Univariate Linear Regression')
    plt.legend()
    plt.show()

    # Calculate and print sum of squares error for the interactive line
    ss_error = np.sum((y - y_pred) ** 2)
    print(f"Sum of squares error (Interactive Line): {ss_error}")

    # Calculate and print sum of squares error for the least squares line
    ss_error_least_squares = np.sum((y - y_pred_least_squares) ** 2)
    print(f"Sum of squares error (Least Squares Line): {ss_error_least_squares}")

    # Display the optimal gradient and intercept
    print(f"Optimal Gradient (Slope): {theta_best[1][0]}")
    print(f"Optimal Intercept: {theta_best[0][0]}")

    # Display the optimal gradient and intercept
    print(f"Interactive Line Gradient (Slope): {m}")
    print(f"Interactive Line Intercept: {c}")
    print(f"Therefore, optimal linear regression with sum of least squares error is : y={theta_best[1][0]}x+{theta_best[0][0]}")

# Create interactive sliders for slope (m) and intercept (c)
interact(plot_regression_line, m=(-10, 10, 0.1), c=(-10, 10, 0.1));
```

In the

1. **Equation for the Line (Interactive Line):**
   $$ y = mx + c $$
   - Where:
     - $ y $ is the dependent variable.
     - $ m $ is the slope of the line (interactive parameter).
     - $ x $ is the independent variable.
     - $ c $ is the y-intercept of the line (interactive parameter).

2. **Normal Equation for Computing Least Squares Regression Line:**
   $$ \theta = (X^T X)^{-1} X^T y $$
   - Where:
     - $ \theta $ is the parameter vector containing the intercept and slope (gradient).
     - $ X $ is the matrix of input features, augmented with a column of ones for the intercept term.
     - $ y $ is the vector of target values.

3. **Equation for the Least Squares Regression Line:**
   $$ y = \theta_1 x + \theta_0 $$
   - Where:
     - $ y $ is the dependent variable.
     - $ \theta_1 $ is the slope (gradient) of the least squares regression line (calculated from the normal equation).
     - $ x $ is the independent variable.
     - $ \theta_0 $ is the y-intercept of the least squares regression line (calculated from the normal equation).

4. **Sum of Squares Error Calculation (for both lines):**
   $$ SSE = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $$
   - Where:
     - $ SSE $ stands for Sum of Squares Error.
     - $ y_i $ is the actual value from the dataset.
     - $ \hat{y}_i $ is the predicted value from the model (either the interactive line or the least squares line).
