# **🛠️ Feature Engineering**

**❓ _What is feature engineering?_**

- Feature engineering is the process of creating, transforming, or selecting variables to improve model performance.

- Good features often contribute more to accuracy than the choice of model itself.

---
## **🛠️ Curse of Dimensionality**

**❓ _What is the curse of dimensionality?_**

- The curse of dimensionality refers to problems that arise when working with high-dimensional data.

- As the number of features increases, the data becomes sparse, distance metrics become less meaningful, and model performance degrades.

- This is why dimensionality reduction or feature selection is often necessary in high-dimensional datasets.

---
## **🛠️ Multicollinearity**

**❓ _What is multicollinearity and why is it a problem?_**

- Multicollinearity occurs when two or more features in a dataset are highly correlated.

- It makes it hard for models like linear regression to estimate the true relationship between features and the target because the model can't distinguish the individual effect of correlated variables.

- It can inflate variance of coefficients, making the model unstable and less interpretable.

---
## **🛠️ Data Leakage**

**❓ _What is data leakage and how do you prevent it?_**

- Data leakage happens when information from outside the training dataset is used to create the model, leading to overly optimistic performance during evaluation.

- A common cause is applying preprocessing (like scaling or imputation) before splitting the data.

- The best way to prevent it is to fit all transformations only on the training set, then apply to test or validation data.

## **🤖 Parametric vs Non-Parametric**

**❓ _What is the difference between parametric and non-parametric models?_**

- Parametric models assume a fixed form for the function and estimate a finite set of parameters (like linear regression).

- Non-parametric models make fewer assumptions about the form and can grow more complex with more data (like KNN or decision trees).

- Parametric models are faster and simpler; non-parametric ones are more flexible but may need more data and computation.

# **🧩 Model Complexity**

---
## **🧩 Bias-Variance Tradeoff**

**❓ _What is the bias-variance tradeoff_**

- The bias-variance tradeoff is about balancing two types of errors in a model.

  - Bias: How far off your predictions are on average from the actual values.
  - Variance: How much your predictions change if the model is trained on different subsets of the data.

- Bias refers to error from overly simplistic assumptions — high bias means the model underfits.

- Variance refers to error from being too sensitive to training data — high variance leads to overfitting.

- A good model finds the sweet spot: low enough bias to capture patterns, but not so low variance that it memorizes noise.

- The goal is to minimize total error by balancing both — not just reducing one at the cost of the other.

---
## **🧩 Overfitting vs Underfitting**

**❓ _What is the difference between overfitting and underfitting?_**

- Overfitting happens when a model learns not just the patterns but also the noise in the training data — it performs well on training data but poorly on unseen data.

- Underfitting is when a model is too simple to capture the underlying patterns — it performs poorly on both training and test data.

- The ideal model generalizes well: it captures the signal without memorizing the noise.

---
## **🧩 Regularization**

**❓ _What is regularization and why is it used?_**

- Regularization is a technique used to prevent overfitting by adding a penalty term to the loss function.

- In L1 (Lasso) regularization, it encourages sparsity by shrinking some coefficients to zero.

- In L2 (Ridge), it shrinks coefficients but keeps all features.

- The idea is to constrain the model complexity for better generalization.

**❓ _What is regularization strength (lambda or alpha) and how does it affect the model?_**

- The regularization parameter controls the penalty applied to model complexity.

- A high value forces the model to shrink coefficients more, potentially underfitting.

- A low value allows more flexibility but risks overfitting.

---
## **🧩 Hyperparameters**

**❓ _What are hyperparameters vs parameters?_**

- Parameters are learned by the model during training (like weights).

- Parameters are what the model learns.

- Hyperparameters are set before training (like learning rate, tree depth) and tuned to improve performance.

- Hyperparameters influence how the model learns.

# **📊 Model Evaluation**

---
## **📊 Cross-Validation**

**❓ _What is cross-validation and why is it important?_**

- Cross-validation is a technique to evaluate a model’s performance more reliably by training and testing it on different subsets of the data.

- The most common form, K-Fold Cross-Validation, helps ensure the model’s performance is not dependent on just one particular train-test split.

- It’s especially important when working with small datasets or tuning hyperparameters.

---
## **📊 ROC**

**❓ _What is AUC-ROC and what does it tell you?_**

- AUC-ROC measures a model’s ability to distinguish between classes across all thresholds.

- AUC closer to 1 means the model ranks positive samples higher than negative ones very well.

- It’s especially useful in binary classification with imbalanced classes.

- When given a random positive and negative sample, a good AUC model will most often than not rank the positive sample higher than the negative sample.
