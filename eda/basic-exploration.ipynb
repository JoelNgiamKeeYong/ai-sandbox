## **└─ Correlation Analysis**

- The heatmap shows the patterns of missingness in your dataset to help identify systematic issues in data collection or preprocessing.

- It helps identify whether missing values in one column are correlated with missing values in other columns.

```python
# Define missing features
missing_columns = ['workclass', 'occupation', 'native-country']

# Calculate missing value correlation
nullity_corr = df[missing_columns].isnull().corr()

# Create a mask for the upper triangle (including diagonal)
mask = np.triu(np.ones_like(nullity_corr, dtype=bool))

# Plot using seaborn
plt.figure(figsize=(8, 4))
sns.heatmap(
    nullity_corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="flare",
    vmin=-1, vmax=1,
    linewidths=0.5,
    cbar=True,
    square=True
)

# Final plot tweaks
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.title("Missing Value Correlation Heatmap (Lower Triangle Only)")
plt.tight_layout()
plt.show();
```
