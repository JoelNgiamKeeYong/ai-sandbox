# **📊📈📉 Data Distributions**

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# **⚖️ Symmetric Distributions**

---
---
## **└─ Normal (Gaussian)**

- Bell-shaped, symmetric, defined by mean & variance.

---
**Human Heights (Adult Males)**

- Typically follows a normal distribution.
- Mean ≈ 175 cm
- Standard Deviation ≈ 7 cm

```python
np.random.seed(42)
data = np.random.normal(loc=175, scale=7, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=40, kde=True, color='mediumseagreen')
plt.title("Human Heights (Adult Males)")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.show()
```

---
**IQ Scores**

- Mean = 100
- Standard Deviation = 15

```python
np.random.seed(42)
data = np.random.normal(loc=100, scale=15, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=40, kde=True, color='steelblue')
plt.title("IQ Scores")
plt.xlabel("IQ")
plt.ylabel("Frequency")
plt.show()
```

---
**Measurement Errors in Scientific Experiments**

- Random experimental errors (due to instruments, environment, etc.) often follow a normal distribution centered at 0.
- Mean = 0
- Standard Deviation = 0.5

```python
np.random.seed(42)
data = np.random.normal(loc=0, scale=0.5, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=40, kde=True, color='tomato')
plt.title("Measurement Errors in Experiments")
plt.xlabel("Error")
plt.ylabel("Frequency")
plt.show()
```

## **└─ Uniform**

- Equal probability across range, flat shape.

---
**🎲 Lottery Numbers**

- In lottery systems where numbers are drawn randomly, each number within the range has an equal chance of being selected.
- Range: 1 to 49
- Real-world assumption: No bias in draw mechanism.

```python
np.random.seed(42)
lottery_numbers = np.random.uniform(1, 49, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(lottery_numbers, bins=40, kde=False, color='orchid')
plt.title("Uniform Distribution: Lottery Numbers (1–49)")
plt.xlabel("Lottery Number")
plt.ylabel("Frequency")
plt.show()
```

---
**⏱️ Arrival Times in a Fixed Window**

- When arrivals are equally likely at any time within a fixed period (e.g., 9:00 to 9:30 AM), the data follows a uniform distribution.
- Range: 9.0 to 9.5 (representing 9:00 to 9:30 AM)

```python
np.random.seed(42)
arrival_times = np.random.uniform(9.0, 9.5, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(arrival_times, bins=40, kde=False, color='skyblue')
plt.title("Uniform Distribution: Arrival Times (9:00–9:30 AM)")
plt.xlabel("Arrival Time (Hours)")
plt.ylabel("Frequency")
plt.show()
```

---
**🔌 Sensor Noise or Analog Jitter**

- Analog sensors may exhibit random noise within a bounded voltage range, especially when no signal is present.
- Range: -0.5 to 0.5 volts
- This uniform distribution represents equal likelihood of small random deviations.

```python
np.random.seed(42)
sensor_noise = np.random.uniform(-0.5, 0.5, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(sensor_noise, bins=40, kde=False, color='orange')
plt.title("Uniform Distribution: Sensor Noise (−0.5V to +0.5V)")
plt.xlabel("Voltage")
plt.ylabel("Frequency")
plt.show()
```

## **└─ Bimodal**

- Two or more distinct peaks/modes.

---
**🧑‍🎓 Exam Scores from Two Distinct Groups**

- Imagine test scores from two groups of students:
  - Students who studied well (mean ≈ 80)
  - Students who didn’t (mean ≈ 50)

- When combined, their scores form a bimodal distribution — two peaks, one for each group.

- This is common in grouped population data (e.g., beginners vs experts).

```python
np.random.seed(42)

# Generate scores for two distinct groups
group1 = np.random.normal(loc=50, scale=5, size=500)  # Low performers
group2 = np.random.normal(loc=80, scale=5, size=500)  # High performers

bimodal_scores = np.concatenate([group1, group2])

plt.figure(figsize=(6, 4))
sns.histplot(bimodal_scores, bins=40, kde=True, color='mediumpurple')
plt.title("Bimodal Distribution: Exam Scores of Two Student Groups")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

---
**🚻 Heights of Adult Males and Females**

- If you plot the heights of both adult males and females together, you'll often get two peaks:
  - Female mean height ≈ 162 cm
  - Male mean height ≈ 175 cm

- This creates a bimodal distribution due to physiological differences between genders.

```python
np.random.seed(42)

# Generate heights for two genders
female_heights = np.random.normal(loc=162, scale=5, size=500)
male_heights = np.random.normal(loc=175, scale=5, size=500)

heights_combined = np.concatenate([female_heights, male_heights])

plt.figure(figsize=(6, 4))
sns.histplot(heights_combined, bins=40, kde=True, color='lightcoral')
plt.title("Bimodal Distribution: Heights of Adult Males and Females")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.show()
```

---
**🛒 Shopping Times: Lunch vs Evening Crowd**

- Shoppers might visit a store during lunch (12–2 PM) or evening (6–8 PM).
- If you combine both windows, the resulting distribution of arrival times shows two clear peaks.

```python
np.random.seed(42)

# Simulate shopping times in hours (24h format)
lunch_shoppers = np.random.normal(loc=13, scale=0.5, size=500)   # Lunch crowd
evening_shoppers = np.random.normal(loc=19, scale=0.5, size=500) # Evening crowd

shopping_times = np.concatenate([lunch_shoppers, evening_shoppers])

plt.figure(figsize=(6, 4))
sns.histplot(shopping_times, bins=40, kde=True, color='mediumseagreen')
plt.title("Bimodal Distribution: Shopping Times (Lunch & Evening)")
plt.xlabel("Time of Day (Hours)")
plt.ylabel("Frequency")
plt.show()
```

## **└─ Multimodal**

- More than one mode or peak.

---
**🍦 Ice Cream Sales by Temperature and Season**

- Ice cream sales spike during hot weekends, school holidays, and festival days.
- These distinct sales peaks can result in a multimodal distribution.

```python
np.random.seed(42)

# Simulate 3 seasonal peaks in sales
peak1 = np.random.normal(loc=100, scale=10, size=300)  # Normal warm days
peak2 = np.random.normal(loc=150, scale=10, size=300)  # Holidays
peak3 = np.random.normal(loc=200, scale=10, size=300)  # Festivals

ice_cream_sales = np.concatenate([peak1, peak2, peak3])

plt.figure(figsize=(6, 4))
sns.histplot(ice_cream_sales, bins=40, kde=True, color='hotpink')
plt.title("Multimodal: Ice Cream Sales Over Different Peak Periods")
plt.xlabel("Units Sold")
plt.ylabel("Frequency")
plt.show()
```

---
**🎧 Listening Habits Across Age Groups**

- Younger, middle-aged, and older adults have different preferred times to stream music.
- If combined, it shows three usage peaks: morning, afternoon, and late night.

```python
np.random.seed(42)

# Simulate listening times (24h format)
young = np.random.normal(loc=22, scale=1, size=300)    # Late night
middle = np.random.normal(loc=15, scale=1, size=300)   # Afternoon
elderly = np.random.normal(loc=9, scale=1, size=300)   # Morning

listening_times = np.concatenate([young, middle, elderly])

plt.figure(figsize=(6, 4))
sns.histplot(listening_times, bins=40, kde=True, color='royalblue')
plt.title("Multimodal: Music Listening Times Across Age Groups")
plt.xlabel("Time of Day (Hours)")
plt.ylabel("Frequency")
plt.show()
```

---
**🛫 Flight Departures by Region**

- Different airports or regions have independent scheduling patterns.
- When aggregated (e.g., across the U.S.), it leads to a multimodal pattern.

```python
np.random.seed(42)

# Simulate flight departures (hour of day)
east_coast = np.random.normal(loc=6, scale=1, size=300)   # Early morning
midwest = np.random.normal(loc=12, scale=1, size=300)     # Midday
west_coast = np.random.normal(loc=18, scale=1, size=300)  # Evening

flight_times = np.concatenate([east_coast, midwest, west_coast])

plt.figure(figsize=(6, 4))
sns.histplot(flight_times, bins=40, kde=True, color='darkorange')
plt.title("Multimodal: Flight Departure Times by Region")
plt.xlabel("Departure Time (Hour)")
plt.ylabel("Frequency")
plt.show()
```

---
**🩸 Health Metric with Threshold (e.g. Blood Sugar, BMI)**

- Example: Values tightly packed with clear upper/lower boundaries (normal, pre-diabetic, diabetic).

```python
np.random.seed(42)
data = np.concatenate([
    np.random.normal(loc=90, scale=5, size=700),   # Normal
    np.random.normal(loc=115, scale=5, size=200),  # Pre-diabetic
    np.random.normal(loc=150, scale=5, size=100)   # Diabetic
])

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=40, kde=True, color='tomato')
plt.title("Multimodal Health Metric (e.g. Blood Glucose)")
plt.xlabel("mg/dL")
plt.ylabel("Frequency")
plt.axvline(100, color='black', linestyle='--', label="Pre-diabetic threshold")
plt.axvline(126, color='red', linestyle='--', label="Diabetic threshold")
plt.legend()
plt.show()
```

# **📉 Skewed Distributions**

---
---
## **└─ Right Skew**

- Positive skew
- Tail stretches to the right (higher values).

---
**💰 Income Distribution**

- Most people earn near the median income, but a few earn significantly more (top 1%), forming a long right tail.
- Common example of a positively skewed or right-skewed distribution.

```python
np.random.seed(42)

income = np.random.exponential(scale=30000, size=1000)  # Mean ~ 30k, right-skewed

plt.figure(figsize=(6, 4))
sns.histplot(income, bins=40, kde=True, color='mediumseagreen')
plt.title("Right-Skewed: Income Distribution")
plt.xlabel("Income ($)")
plt.ylabel("Frequency")
plt.show()
```

---
**📞 Customer Service Call Durations**

- Most calls are short, but a few last significantly longer (angry customers, tech issues), resulting in a right-skew.
- The distribution is compressed on the left with a long tail to the right.

```python
np.random.seed(42)

call_durations = np.random.gamma(shape=2, scale=2, size=1000) * 5  # Multiply to increase spread

plt.figure(figsize=(6, 4))
sns.histplot(call_durations, bins=40, kde=True, color='tomato')
plt.title("Right-Skewed: Call Duration (Minutes)")
plt.xlabel("Call Length")
plt.ylabel("Frequency")
plt.show()
```

---
**🛒 Number of Items Per Transaction**

- In many retail settings, most customers buy only a few items, while a small number bulk buy (e.g. warehouse stores).
- Creates a right-skewed distribution with many small values and few large ones.

```python
np.random.seed(42)

items_bought = np.random.poisson(lam=2, size=1000) + np.random.binomial(n=10, p=0.1, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(items_bought, bins=range(0, max(items_bought)+1), kde=False, color='goldenrod')
plt.title("Right-Skewed: Items Purchased per Transaction")
plt.xlabel("Number of Items")
plt.ylabel("Frequency")
plt.show()
```

---
---
## **└─  Left Skew**

- Negative skew
- Tail stretches to the left (lower values).

---
**📚 Exam Completion Time (in minutes)**

- Most students finish near the maximum allowed time, but a few finish very quickly, causing a left skew.

```python
np.random.seed(42)

# Simulate exam completion times: left skew (most finish late, some finish early)
completion_time = 100 - np.random.exponential(scale=10, size=900)  # Skewed left
completion_time = np.clip(completion_time, 0, 100)  # Limit range to 0-100

plt.figure(figsize=(6, 4))
sns.histplot(completion_time, bins=40, kde=True, color='teal')
plt.title("Left Skew: Exam Completion Times (minutes)")
plt.xlabel("Completion Time (minutes)")
plt.ylabel("Frequency")
plt.show()
```

---
**🧓 Age at Retirement (years)**

- Most people retire close to the maximum retirement age, but some retire early, creating a left skew.

```python
np.random.seed(42)

# Simulate retirement ages: left skew (most retire late, some early)
retirement_age = 70 - np.random.gamma(shape=2, scale=3, size=900)
retirement_age = np.clip(retirement_age, 40, 70)  # Clip realistic ages

plt.figure(figsize=(6, 4))
sns.histplot(retirement_age, bins=40, kde=True, color='crimson')
plt.title("Left Skew: Age at Retirement (years)")
plt.xlabel("Retirement Age (years)")
plt.ylabel("Frequency")
plt.show()
```

---
**💵 Days Taken to Pay Off Loan**

- Many people take almost the full loan period to pay off, but some pay off much earlier, producing left skew.

```python
np.random.seed(42)

# Simulate days to pay off loan: left skew
loan_payoff = 365 - np.random.chisquare(df=4, size=900) * 20
loan_payoff = np.clip(loan_payoff, 0, 365)

plt.figure(figsize=(6, 4))
sns.histplot(loan_payoff, bins=40, kde=True, color='navy')
plt.title("Left Skew: Days to Pay Off Loan")
plt.xlabel("Days to Pay Off Loan")
plt.ylabel("Frequency")
plt.show()
```

# **🐍 Tailed Distributions**

---
---
## **└─ Long-Tailed**

- Heavy tailed
- Extreme values occur more frequently than normal (e.g., Pareto, Cauchy).

---
**🎵 Customer Purchase Amounts ($)**

- Most customers spend relatively small amounts, but a few make very large purchases.
- This creates a long tail on the right, representing rare but very high spenders.
- Such distributions are common in retail and e-commerce.

```python
np.random.seed(42)

# Simulate purchase amounts with a long right tail
purchase_amounts = np.random.pareto(a=3, size=900) * 50

plt.figure(figsize=(6, 4))
sns.histplot(purchase_amounts, bins=40, kde=True, color='dodgerblue')
plt.title("Long-Tailed: Customer Purchase Amounts ($)")
plt.xlabel("Purchase Amount ($)")
plt.ylabel("Frequency")
plt.show()
```

---
**🎯 Website Session Durations (seconds)**

- Most sessions are short, but a small number last a very long time.
- This right-skewed long tail indicates occasional highly engaged users.
- Understanding this helps optimize site experience for both casual and power users.

```python
np.random.seed(42)

# Simulate session durations with a long right tail
session_durations = np.random.lognormal(mean=3, sigma=1, size=900)

plt.figure(figsize=(6, 4))
sns.histplot(session_durations, bins=40, kde=True, color='mediumseagreen')
plt.title("Long-Tailed: Website Session Durations (seconds)")
plt.xlabel("Session Duration (seconds)")
plt.ylabel("Frequency")
plt.show()
```

---
**⏳ Time Between Failures (hours)**

- Most failures happen quickly, but some machines run a very long time before failing.
- The heavy long tail on the right shows the variability in machine reliability.
- This distribution is critical for maintenance scheduling and reliability engineering.

```python
np.random.seed(42)

# Simulate time between failures with a long right tail
time_between_failures = np.random.weibull(a=1.5, size=900) * 100

plt.figure(figsize=(6, 4))
sns.histplot(time_between_failures, bins=40, kde=True, color='darkorange')
plt.title("Long-Tailed: Time Between Failures (hours)")
plt.xlabel("Time Between Failures (hours)")
plt.ylabel("Frequency")
plt.show()
```

---
---
## **└─ Light-Tailed**

- Less extreme values than normal.

---
**🎵 Customer Wait Times at a Coffee Shop (minutes)**

- Most customers are served quickly, with wait times tightly clustered around the average.
- Very long wait times are rare, so the tails drop off sharply.
- This reflects efficient service and consistent processing times.

---

# **🎯 Discrete Distributions**

---
---
## **└─ Binomial / Bernoulli**

- This distribution models the probability of a specific number of successes in a fixed number of independent trials, where each trial has only two possible outcomes (success or failure).

- For example, flipping a coin multiple times and counting the number of heads. 

- Outcomes 0 or 1 (success/failure).

---
**🖱️ Ad Clicks per User (out of 10 ads)**

- Most users click on around 3 ads, reflecting the expected success rate (30%).
- It's rare for users to click on all or none of the ads.
- This distribution is typical in marketing experiments measuring user response behavior under a fixed number of trials.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
# 1000 users, each sees 10 ads with 30% click probability
data = np.random.binomial(n=10, p=0.3, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=range(12), discrete=True, color='lightgreen')
plt.title("Binomial Distribution (n=10, p=0.3)")
plt.xlabel("Number of Clicks")
plt.ylabel("Frequency")
plt.show()
```

---
**🏭 Items Passing Quality Check (per batch of 20)**

- Most batches have 18 to 20 items passing QC, indicating a very high quality rate.
- Rare to see batches with fewer than 18 passed items.
- This reflects a consistent manufacturing process with tight control over defects.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(24)
# Each batch has 20 items, with 95% probability of passing QC
data = np.random.binomial(n=20, p=0.95, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=range(18, 22), discrete=True, color='tomato')
plt.title("Binomial Distribution (n=20, p=0.95)")
plt.xlabel("Number of Items Passed QC")
plt.ylabel("Frequency")
plt.show()
```

---
**📮 Number of People Responding to a Survey (out of 15 contacted)**

- Most people don’t respond, so 0 to 2 replies are most common.
- A few cases show higher responses, but they're rare.
- This distribution is typical when dealing with cold emails or low engagement surveys.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(7)
# 15 people contacted, with 10% probability of responding
data = np.random.binomial(n=15, p=0.1, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=range(8), discrete=True, color='orchid')
plt.title("Binomial Distribution (n=15, p=0.1)")
plt.xlabel("Number of Responses")
plt.ylabel("Frequency")
plt.show()
```

## **└─ Poisson**

- This distribution models the probability of a certain number of events occurring within a fixed interval of time or space, given a known average rate of occurrence.

- For example, counting the number of cars passing a certain point on a highway in an hour. 

- Counts of rare events over fixed interval.

---
**🏦 Customer Arrivals at a Bank (per minute)**

- Most minutes have 2–4 customer arrivals.
- Occasionally, a minute is busier or quieter, but extremes are rare.
- This shows a steady flow of customers in a typical business setting.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(1)
# Average of 3 customers per minute
data = np.random.poisson(lam=3, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=range(10), discrete=True, color='steelblue')
plt.title("Poisson Distribution (λ=3)")
plt.xlabel("Customers per Minute")
plt.ylabel("Frequency")
plt.show()
```

---
**📘 Typos on Each Page of a Book**

- Most pages have no typos or just one.
- It's rare to see more than 2 typos on a page.
- This distribution is common in quality-controlled content where errors are occasional but random.

```python
np.random.seed(2)
# Average of 1 typo per page
data = np.random.poisson(lam=1, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=range(6), discrete=True, color='darkorange')
plt.title("Poisson Distribution (λ=1)")
plt.xlabel("Typos per Page")
plt.ylabel("Frequency")
plt.show()
```

---
**📞 Calls Received by a Call Center (per hour)**

- Most hours receive between 8–12 calls.
- Occasionally, there's a surge or lull, but it remains centered around 10.
- This pattern reflects typical random fluctuations in service demand.

```python
np.random.seed(3)
# Average of 10 calls per hour
data = np.random.poisson(lam=10, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=range(20), discrete=True, color='mediumseagreen')
plt.title("Poisson Distribution (λ=10)")
plt.xlabel("Calls per Hour")
plt.ylabel("Frequency")
plt.show()
```

# **🏷️ Categorical Distributions**

- Categories without order (e.g., colors).

---
**🍨 Favorite Ice Cream Flavors**

- Chocolate and vanilla are top favorites among customers.
- Flavors like mint have smaller fan bases, though still present.
- The distribution has no order — all flavors are equally valid categories.

## **└─ Zero-Inflated**

- Excess zeros compared to expected distribution.

---
**🦷 Number of Cavities in a Dental Study**

- Most people don’t have cavities, resulting in a large number of zeros.
- Among those who do, the number of cavities follows a count-like distribution (e.g., Poisson).
- This reflects the fact that good dental hygiene leads to a spike in "zero cavities", while some still get multiple.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Simulate data: 70% have 0 cavities, 30% follow Poisson
np.random.seed(42)
cavities = np.where(np.random.rand(1000) < 0.7, 0, np.random.poisson(2, 1000))

# Plot
plt.figure(figsize=(6, 4))
sns.histplot(cavities, bins=range(10), discrete=True)
plt.title("Number of Cavities (Zero-Inflated Poisson)")
plt.xlabel("Number of Cavities")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

---
**🚲 Bicycle Accidents per Commuter per Year**

- Many commuters never get into accidents, leading to an overrepresentation of zero counts.
- Those who do tend to have one or two minor incidents per year.
- This mirrors real-world safety patterns: accidents are rare but follow a pattern when they do occur.

```python
# Simulate data: 80% have 0 accidents, 20% follow Poisson
accidents = np.where(np.random.rand(1000) < 0.8, 0, np.random.poisson(1.5, 1000))

# Plot
plt.figure(figsize=(6, 4))
sns.histplot(accidents, bins=range(8), discrete=True, color='mediumseagreen')
plt.title("Bicycle Accidents per Year (Zero-Inflated Poisson)")
plt.xlabel("Number of Accidents")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

---
**💡 Zero-Heavy User Activity (e.g., Logins, Workouts, or App Usage)**

- Many users have zero activity in a given time window — for example, they didn't log in or exercise that week.
- Among active users, behavior follows a typical Poisson-like usage pattern (e.g., 1–5 logins or workouts).
- This distribution reflects real-world engagement patterns, where a small subset of users are highly active, while the majority remain dormant or sporadic.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Simulate: 60% active users follow Poisson, 40% have zero usage
np.random.seed(42)
data = np.random.poisson(lam=2, size=1000)
data[np.random.rand(1000) < 0.4] = 0  # 40% non-users

# Plot
plt.figure(figsize=(6, 4))
sns.histplot(data, bins=range(data.max() + 2), discrete=True)
plt.title("Zero-Heavy Usage Distribution (e.g., Workouts or Logins)", fontsize=13)
plt.xlabel("Activity Count")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## **└─ Mixture**

- Combination of multiple distributions.

---
**📊 E-commerce Purchase Amounts (Low-value vs High-value Customers)**

- Most customers make low-value purchases, but a small group of high-value customers make large purchases.
- This creates a right-skewed bimodal distribution.
- Useful for customer segmentation (e.g., regular vs premium spenders).

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(1)
low_spenders = np.random.normal(30, 10, 900)
high_spenders = np.random.normal(150, 30, 100)
amounts = np.concatenate([low_spenders, high_spenders])

plt.figure(figsize=(6, 4))
sns.histplot(amounts, bins=40, kde=True)
plt.title("E-commerce Purchase Amounts (Mixture of Spend Levels)")
plt.xlabel("Purchase Amount ($)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

---
**🧠 Student Exam Scores (Prepared vs Unprepared Students)**

- Some students studied well, while others didn’t — leading to a bimodal distribution.
- The two peaks reflect distinct subgroups with different performance levels.
- Helps identify intervention needs in education.

```python
np.random.seed(2)
unprepared = np.random.normal(50, 10, 600)
prepared = np.random.normal(80, 5, 400)
scores = np.concatenate([unprepared, prepared])

plt.figure(figsize=(6, 4))
sns.histplot(scores, bins=30, kde=True)
plt.title("Student Exam Scores (Prepared vs Unprepared)")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

---
**🏥 Hospital Stay Duration (Minor vs Major Illnesses)**

- Most patients have short stays (e.g., for minor conditions), but a few have very long stays due to severe illnesses or complications.
- Results in a mixture of short and long duration peaks, often skewed.
- Valuable in resource and bed management.

```python
np.random.seed(3)
short_stay = np.random.poisson(2, 800)
long_stay = np.random.poisson(10, 200)
stay_duration = np.concatenate([short_stay, long_stay])

plt.figure(figsize=(6, 4))
sns.histplot(stay_duration, bins=range(stay_duration.max()+1), discrete=True)
plt.title("Hospital Stay Duration (Mixture: Minor vs Major Illness)")
plt.xlabel("Days Admitted")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## **└─ Truncated distributions**

- Normal or others cut off at some bounds.

---
**🛑 Truncated Normal Distribution (Height Restricted to Adult Range)**

- Modeling adult heights limited between 150 cm and 190 cm.
- Original normal distribution truncated to exclude extreme values outside this range.
- Useful to model realistic human height data without outliers.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import truncnorm

np.random.seed(42)
# Parameters for truncnorm: a, b are lower and upper bounds in standard deviations
lower, upper = (150 - 170) / 10, (190 - 170) / 10  # mean=170, sd=10
data = truncnorm.rvs(lower, upper, loc=170, scale=10, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=30, kde=True, color='skyblue')
plt.title("Truncated Normal: Adult Height (150cm to 190cm)")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

---
**🛑 Truncated Exponential Distribution (Customer Wait Time, Max 10 Minutes)**

- Modeling wait times truncated to a maximum of 10 minutes.
- Typical for customer service where longer waits are impossible or censored.
- Reflects realistic bounds on service experience.

```python
np.random.seed(24)
# Truncate exponential at 10 (right truncation)
upper = 10
data = truncnorm.rvs(a=0, b=(upper - 2)/2, loc=2, scale=2, size=1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=30, kde=True, color='mediumseagreen')
plt.title("Truncated Exponential: Customer Wait Time (max 10 mins)")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

---
**🛑 Truncated Uniform Distribution (Age Range Restricted)**

- Uniform distribution of ages truncated between 20 and 50 years.
- Useful to model survey data limited to a certain demographic.
- Shows a flat distribution but restricted to a bounded interval.

```python
np.random.seed(99)
lower, upper = 20, 50
data = np.random.uniform(lower, upper, 1000)

plt.figure(figsize=(6, 4))
sns.histplot(data, bins=30, kde=False, color='coral')
plt.title("Truncated Uniform: Age Range 20 to 50")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## **└─ Capped Distribution**

- Mass at Upper Cap
- Survey scores out of 10, credit score bands, rating limits.

---
**🧢 Capped Normal Distribution (Income Capped at $100,000)**

- Simulated income data from a normal distribution but capped at $100,000.
- Reflects a dataset where income beyond $100,000 is either not recorded or limited.
- Useful for modeling censored financial data.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
data = np.random.normal(loc=60000, scale=20000, size=1000)
data_capped = np.clip(data, a_min=None, a_max=100000)

plt.figure(figsize=(6, 4))
sns.histplot(data_capped, bins=30, kde=True, color='skyblue')
plt.title("Capped Normal Distribution: Income capped at $100,000")
plt.xlabel("Income ($)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

---
**🧢 Capped Poisson Distribution (Number of Sales Capped at 10)**

- Counts of sales per day generated by a Poisson distribution, but capped at 10.
- Reflects reporting limits or business logic that ignores very high counts.
- Useful to model capped event counts.

```python
np.random.seed(24)
data = np.random.poisson(lam=5, size=1000)
data_capped = np.clip(data, a_min=None, a_max=10)

plt.figure(figsize=(6, 4))
sns.histplot(data_capped, bins=range(12), discrete=True, color='mediumseagreen')
plt.title("Capped Poisson Distribution: Sales capped at 10")
plt.xlabel("Number of Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

---
**🧢 Capped Uniform Distribution (Ratings Capped at 4 Stars)**

- Uniform ratings between 1 and 5 but capped at 4 stars.
- Reflects a system where max rating is limited to 4 due to policy or scale change.
- Useful to model bounded rating data.

```python
np.random.seed(99)
data = np.random.uniform(low=1, high=5, size=1000)
data_capped = np.clip(data, a_min=None, a_max=4)

plt.figure(figsize=(6, 4))
sns.histplot(data_capped, bins=30, kde=False, color='coral')
plt.title("Capped Uniform Distribution: Ratings capped at 4 stars")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
