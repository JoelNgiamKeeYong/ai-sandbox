| Method                                 | Type of Variables                | Purpose                                              | When to Use                                                                                    | Notes                                                                                               |
| -------------------------------------- | -------------------------------- | ---------------------------------------------------- | ---------------------------------------------------------------------------------------------- | --------------------------------------------------------------------------------------------------- |
| **Pearson’s Correlation**              | Numerical ↔ Numerical            | Measure linear correlation                           | When checking for multicollinearity, or selecting numeric features related to a numeric target | Assumes linear relationship and normal distribution. Values range from -1 to 1.                     |
| **Chi-Squared Test**                   | Categorical ↔ Categorical        | Test independence / association                      | When assessing relationship between categorical feature and target (e.g., in classification)   | Requires sufficient frequency counts. Use only with non-negative integers.                          |
| **ANOVA (F-test)**                     | Numerical ↔ Categorical (target) | Compare means across groups                          | When checking which numerical features differ significantly across target classes              | Used in `SelectKBest` for classification with numeric features. Assumes normality & equal variance. |
| **LDA (Linear Discriminant Analysis)** | Numerical ↔ Categorical (target) | Supervised dimensionality reduction / classification | When reducing dimensionality for classification or as a classifier itself                      | Maximizes class separation. Assumes normally distributed features and equal class covariance.       |
