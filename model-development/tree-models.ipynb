# 🤖🧠🌴 **Tree Models**

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    r2_score, mean_squared_error,
    accuracy_score, precision_score, recall_score, f1_score
)
```

```python
# === 🟨 Regression Dataset: Diabetes
diabetes = load_diabetes()
X_reg = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
y_reg = diabetes.target
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)
```

```python
# === 🟦 Classification Dataset: Breast Cancer
cancer = load_breast_cancer()
X_class = pd.DataFrame(cancer.data, columns=cancer.feature_names)
y_class = cancer.target
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_class, y_class, test_size=0.2, random_state=42)
```

```python
# Pipeline for model evaluation
def evaluate_model(model, is_classification_task=False):
    """
    Fits and evaluates a model with standard metrics for regression or classification.
    
    Parameters:
        model: model instance
        is_classification_task = True for 'classification' and False for 'regression'
    """
    model_name = type(model).__name__

    if is_classification_task == False:
        X_train = X_train_r
        X_test = X_test_r
        y_train = y_train_r
        y_test = y_test_r

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        r2 = r2_score(y_test, y_pred)
        rmse = np.sqrt(mean_squared_error(y_test, y_pred))
        print(f"✔️ {model_name:<20}: R² = {r2:.4f} | RMSE = {rmse:.2f}")
    
    elif is_classification_task == True:
        X_train = X_train_c
        X_test = X_test_c
        y_train = y_train_c
        y_test = y_test_c

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        acc = accuracy_score(y_test, y_pred)
        prec = precision_score(y_test, y_pred, zero_division=0)
        rec = recall_score(y_test, y_pred, zero_division=0)
        f1 = f1_score(y_test, y_pred, zero_division=0)
        print(f"✔️ {model_name:<20}: Acc = {acc:.4f} | Precision = {prec:.4f} | Recall = {rec:.4f} | F1 = {f1:.4f}")

    else:
        raise ValueError("task must be either 'regression' or 'classification'")
```

## 🔖 **Decision Tree Regressor**

```python
from sklearn.tree import DecisionTreeRegressor
dt_model = DecisionTreeRegressor(random_state=42)
evaluate_model(dt_model, is_classification_task=False)
```
