# 🤖🧠🛠️ **Linear Models**

```python
from sklearn.datasets import load_diabetes, load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
# For regression
X_reg, y_reg = load_diabetes(return_X_y=True)
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

# Standardize features for regression
scaler_reg = StandardScaler().fit(X_reg_train)
X_reg_train_scaled = scaler_reg.transform(X_reg_train)
X_reg_test_scaled = scaler_reg.transform(X_reg_test)
```

```python
# For classification
X_clf, y_clf = load_iris(return_X_y=True)
X_clf_train, X_clf_test, y_clf_train, y_clf_test = train_test_split(X_clf, y_clf, test_size=0.2, random_state=42)

# Standardize features for classification
scaler_clf = StandardScaler().fit(X_clf_train)
X_clf_train_scaled = scaler_clf.transform(X_clf_train)
X_clf_test_scaled = scaler_clf.transform(X_clf_test)
```

---
---
---
# **🔖 Regression Models**

---
---
## 🔖 **Linear Regression**

```python
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()
lr_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = lr_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ Linear Regression - MSE: {mse:.4f}")
```

---
---
## 🔖 **Ridge Regression**

```python
from sklearn.linear_model import Ridge
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = ridge_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ Ridge Regression - MSE: {mse:.4f}")
```

---
---
## 🔖 **Lasso Regression**

```python
from sklearn.linear_model import Lasso
lasso_model = Lasso(alpha=0.1)
lasso_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = lasso_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ Lasso Regression - MSE: {mse:.4f}")
```

---
---
## 🔖 **ElasticNet**

```python
from sklearn.linear_model import ElasticNet
elastic_net_model = ElasticNet(alpha=0.1, l1_ratio=0.5)
elastic_net_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = elastic_net_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ ElasticNet - MSE: {mse:.4f}")
```

---
---
## 🔖 **LARS**

```python
from sklearn.linear_model import Lars
lars_model = Lars(n_nonzero_coefs=10)
lars_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = lars_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ LARS - MSE: {mse:.4f}")
```

---
---
## 🔖 **LassoLARS**

```python
from sklearn.linear_model import LassoLars
lasso_lars_model = LassoLars(alpha=0.1)
lasso_lars_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = lasso_lars_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ Lasso LARS - MSE: {mse:.4f}")
```

---

```python
from sklearn.linear_model import OrthogonalMatchingPursuit  
omp_model = OrthogonalMatchingPursuit(n_nonzero_coefs=10)
omp_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = omp_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ OMP - MSE: {mse:.4f}")
```

```python
from sklearn.linear_model import BayesianRidge  
bayesian_ridge_model = BayesianRidge()
bayesian_ridge_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = bayesian_ridge_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ Bayesian Ridge - MSE: {mse:.4f}")
```

```python
from sklearn.linear_model import SGDRegressor  
sgd_regressor_model = SGDRegressor(max_iter=1000, tol=1e-3)
sgd_regressor_model.fit(X_reg_train_scaled, y_reg_train)
y_pred = sgd_regressor_model.predict(X_reg_test_scaled)
mse = mean_squared_error(y_reg_test, y_pred)
print(f"✔️ SGD Regressor - MSE: {mse:.4f}")
```

# **🏷️ Classification Models**

---

## 🏷️ **Logistic Regression (lbfgs)**

```python
from sklearn.linear_model import LogisticRegression  
lr_model = LogisticRegression(max_iter=1000, solver='lbfgs')
lr_model.fit(X_clf_train_scaled, y_clf_train)
y_pred = lr_model.predict(X_clf_test_scaled)
acc = accuracy_score(y_clf_test, y_pred)
print(f"Logistic Regression - Accuracy: {acc:.4f}")
```
