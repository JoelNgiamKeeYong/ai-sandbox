```python
import pandas as pd
import requests
import io
import sqlite3
import json
import zipfile
from bs4 import BeautifulSoup
from sqlalchemy import create_engine
```

## **📥 CSV from a URL**

```python
# Download CSV from URL
csv_url = "https://people.sc.fsu.edu/~jburkardt/data/csv/hw_200.csv"
df_csv = pd.read_csv(csv_url)

print("✅ CSV DataFrame Loaded")
df_csv.head()
```

## **📥 Excel File from Local or Web**

```python
excel_url = "https://www.contextures.com/SampleData.zip"

# This ZIP file contains an Excel file: `SampleData.xlsx`
# Let's extract and read it

# Download and extract Excel file from zip
r = requests.get(excel_url)
z = zipfile.ZipFile(io.BytesIO(r.content))
excel_file = z.open("SampleData.xlsx")

# Load Excel file into DataFrame
df_excel = pd.read_excel(excel_file)

print("✅ Excel DataFrame Loaded")
df_excel.head()
```

## **📥 JSON Data (API or File)**

```python
# 🌐 JSON from API
json_url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(json_url)
data_json = response.json()

df_json = pd.DataFrame(data_json)
print("✅ JSON DataFrame Loaded")
df_json.head()
```

## **📥 Web Scraping with BeautifulSoup**

```python
# 🌍 HTML Table from Wikipedia
wiki_url = "https://en.wikipedia.org/wiki/List_of_countries_by_GDP_(nominal)"
tables = pd.read_html(wiki_url)

# Usually first or second table is the desired one
df_wiki = tables[1]
print("✅ HTML Table Loaded")
df_wiki.head()
```

## **📥 From a Local SQLite Database**

```python
# 🗃️ Create Example SQLite Database
conn = sqlite3.connect("../data/example.db")
cur = conn.cursor()

cur.execute("CREATE TABLE IF NOT EXISTS users (id INTEGER, name TEXT, age INTEGER)")
cur.execute("INSERT INTO users VALUES (1, 'Alice', 30), (2, 'Bob', 24), (3, 'Charlie', 29)")
conn.commit()

# Load into DataFrame
df_sqlite = pd.read_sql("SELECT * FROM users", conn)
conn.close()

print("✅ SQLite DataFrame Loaded")
df_sqlite.head()
```

## **📥 From a SQL Database via SQLAlchemy**

```python
# 🧬 Example: In-memory SQLite with SQLAlchemy
engine = create_engine('sqlite:///:memory:')
df_sqlite.to_sql('users', engine, index=False)

# Read from SQLAlchemy engine
df_from_engine = pd.read_sql("SELECT * FROM users", engine)
print("✅ SQLAlchemy DataFrame Loaded")
df_from_engine.head()
```
