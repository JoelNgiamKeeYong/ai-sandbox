# **Normallized Discounted Cumulative Gain (nDCG)**

When to Use nDCG?

🏆 Ranking tasks (e.g., search engine results, recommender systems)

📈 Evaluation for learning-to-rank algorithms like XGBoostRanker, LightGBM ranker

Ideal when relevance is graded (e.g., ratings 0–3, not just binary).

```python
from sklearn.metrics import ndcg_score

# y_true and y_score should be 2D arrays (samples x labels)
y_true = [[3, 2, 3, 0, 1, 2]]
y_score = [[0.9, 0.8, 0.75, 0.4, 0.2, 0.1]]

ndcg_score(y_true, y_score, k=3)
```

y_true (2D list or array):
- Each row represents a query.
- Each element in the row is the true relevance of an item for that query.
- Values are usually graded relevance scores, e.g., 3 = highly relevant, 2 = relevant, 1 = slightly relevant, 0 = not relevant

y_score (2D list or array):
- Each row represents the model's predicted scores for the same items.
- These are predicted ranking scores (e.g., logits, probabilities, model outputs — not necessarily in 0-1 range).
- The higher the score, the higher the model ranks that item.

# **Mean Average Precision (MAP)**

- MAP evaluates ranking quality by averaging the precision at each relevant position in a list. Then it averages this across all queries.

Let's assume two queries with binary relevance:

```python
# Relevance scores: 1 = relevant, 0 = not relevant
# Predicted rankings per query
y_true = [
    [1, 0, 1, 0, 1],  # Query 1
    [0, 0, 1, 1, 0]   # Query 2
]
```

```python
def mean_average_precision_at_k(y_true_list, k=5):
    """
    Compute Mean Average Precision at k (MAP@k).

    Parameters:
    - y_true_list: List of lists with binary relevance (1 for relevant, 0 for not relevant).
    - k: Integer. Cutoff rank.

    Returns:
    - MAP@k: Float
    """
    def average_precision(y_true, k):
        y_true = y_true[:k]
        score = 0.0
        num_hits = 0.0
        for i, rel in enumerate(y_true):
            if rel:
                num_hits += 1
                score += num_hits / (i + 1.0)
        return score / min(sum(y_true), k) if sum(y_true) > 0 else 0.0

    average_precisions = [
        average_precision(y_true, k) for y_true in y_true_list
    ]
    
    mean_ap = sum(average_precisions) / len(y_true_list)
    
    return mean_ap
```

```python
map = mean_average_precision_at_k(y_true, k=5)
print(f"Mean Average Precision at k=5: {map:.4f}")
```

# **Mean Reciprocal Rank (MRR)**

- MRR evaluates the rank position of the first relevant item in each query result. 

```python
y_true = [
    [0, 0, 1, 0],  # first relevant item at rank 3 → reciprocal = 1/3
    [1, 0, 0, 0],  # first relevant item at rank 1 → reciprocal = 1/1
    [0, 0, 0, 0]   # no relevant item → reciprocal = 0
]
```

```python
def mean_reciprocal_rank_at_k(y_true_list, k=5):
    """
    Compute Mean Reciprocal Rank at k (MRR@k).

    Parameters:
    - y_true_list: List of lists with binary relevance (1 for relevant, 0 for not relevant).
    - k: Integer. Cutoff rank.

    Returns:
    - MRR@k: Float
    """
    def reciprocal_rank(y_true, k):
        y_true = y_true[:k]
        for i, rel in enumerate(y_true):
            if rel:
                return 1.0 / (i + 1.0)
        return 0.0

    reciprocal_ranks = [
        reciprocal_rank(y_true, k) for y_true in y_true_list
    ]

    mean_rr = sum(reciprocal_ranks) / len(y_true_list)

    return mean_rr
```

```python
mrr = mean_reciprocal_rank_at_k(y_true, k=5)
print(f"Mean Reciprocal Rank at k=5: {mrr:.4f}")
```
